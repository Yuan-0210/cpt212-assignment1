```python
# Graph of Algorithm 1 (NumberSort.java)
import matplotlib.pyplot as plt
import numpy as np

# Data from the table
n_values = np.array([8, 16, 24, 32, 40, 48, 56, 64, 72, 80])
primitive_operations = np.array([821, 1277, 1733, 2189, 2645, 3101, 3557, 4013, 4469, 4925])

# Plotting the graph
plt.figure(figsize=(8, 5))
plt.plot(n_values, primitive_operations, marker='o', linestyle='-', color='b')

# Adding title and labels
plt.title("Number of Primitive Operations vs Input Size (n)")
plt.xlabel("Input Size (n)")
plt.ylabel("Number of Primitive Operations")

# Adding grid and setting x-axis ticks for better readability
plt.grid(True)
plt.xticks(n_values)
plt.tight_layout()

# Display the plot
plt.show()
```
