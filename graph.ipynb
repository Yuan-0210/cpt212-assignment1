```python
# Graph of Algorithm 1

# n=8, operationCounts=1832; n=16, operationCounts=2866; n=24, operationCounts=3900 , n=32, operationCounts=4932, n=40, operationCounts=5964.
# n=48. operationCounts=6996 n=56, operationCounts=8028, n=64, operationCounts=9060, , n=72, operationCounts=10092, n=80, operationCounts=11124
# operationCounts from n=16 to n=24 increases by 1034, the rest has difference of 1032

import matplotlib.pyplot as plt
import numpy as np

# Define the function representing the number of primitive operations
def operationCounts(n):
    return 129 * n + 804

# Generate values for n from 8 to 80, stepping by 8
n_values = np.arange(8, 81, 8)
operationCounts_values = operationCounts(n_values)

# Plotting the graph
plt.figure(figsize=(8, 5))
plt.plot(n_values, operationCounts_values, marker='o', linestyle='-', color='b')

# Adding title and labels with updates
plt.title("Number of Primitive Operations vs Input Size, n")
plt.xlabel("n")
plt.ylabel("Number of Primitive Operations")

# Adding grid and setting x-axis ticks for better readability
plt.grid(True)
plt.xticks(n_values)
plt.tight_layout()

# Display the plot
plt.show()



```

```python
# Graph of Algorithm 2

# n=8, operationCounts=3454, n=16, operationCounts=4726, n=24, operationCounts=5998, n=32, operationCounts=7270, n=40, operationCounts=8542
# n=48, operationCounts=9814, n=56, operationCounts=11071, n=64, operationCounts=12343, n=72, operationCounts=13615, n=80, operationCounts=14887 
# operationCounts from n=48 to n=56 increases by 1257, the rest has difference of 1272

import matplotlib.pyplot as plt
import numpy as np

# Define the function representing the number of primitive operations
def operationCounts(n):
    return 159 * n + 2182 - 15 * (n // 56)

# Generate values for n from 8 to 80, stepping by 8
n_values = np.arange(8, 81, 8)
operationCounts_values = operationCounts(n_values)

# Plotting the graph
plt.figure(figsize=(8, 5))
plt.plot(n_values, operationCounts_values, marker='o', linestyle='-', color='b')

# Adding title and labels with updates
plt.title("Number of Primitive Operations vs Input Size, n")
plt.xlabel("n")
plt.ylabel("Number of Primitive Operations")

# Adding grid and setting x-axis ticks for better readability
plt.grid(True)
plt.xticks(n_values)
plt.tight_layout()

# Display the plot
plt.show()
```
