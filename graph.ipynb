```python
# Graph of Algorithm 1 (NumberSort.java)
import matplotlib.pyplot as plt
import numpy as np

# Define the function representing the number of primitive operations
def counter(n):
    return 57 * n + 365

# Generate values for n from 8 to 80, stepping by 8
n_values = np.arange(8, 81, 8)
counter_values = counter(n_values)

# Plotting the graph
plt.figure(figsize=(8, 5))
plt.plot(n_values, counter_values, marker='o', linestyle='-', color='b')

# Adding title and labels with updates
plt.title("Number of Primitive Operations vs Input Size (n)")
plt.xlabel("n")
plt.ylabel("Number of Primitive Operations")

# Adding grid and setting x-axis ticks for better readability
plt.grid(True)
plt.xticks(n_values)
plt.tight_layout()

# Display the plot
plt.show()
```
